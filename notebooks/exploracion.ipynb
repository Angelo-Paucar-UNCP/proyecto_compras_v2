```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if (ROOT / 'src').exists():
    pass
elif (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

from datos import cargar_datos, preparar_y_dividir_datos

df = cargar_datos()
print(df.shape)
print(df.dtypes)
df.head()
```

```python
print('Nulos por columna:')
print(df.isna().sum())
print('\nDesbalance de Revenue:')
print(df['Revenue'].value_counts(normalize=True).round(4))
```

## Calidad y fugas

- No hay nulos en la fuente UCI.
- Clase positiva (compra) ≈ 15.5 %: la accuracy engaña.
- `PageValues` se conoce durante la sesión (valor medio de páginas ya visitadas), no después del pago. Se conserva.
- No usamos el propio `Revenue` como entrada.
- El preprocesador se ajusta solo con entrenamiento.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df['Revenue'].value_counts().plot(kind='bar', ax=axes[0], color=['#94a3b8', '#2563eb'])
axes[0].set_title('Compras vs no compras')
df.boxplot(column='PageValues', by='Revenue', ax=axes[1])
axes[1].set_title('PageValues según compra')
plt.suptitle('')
plt.tight_layout()
Path(ROOT / 'docs').mkdir(exist_ok=True)
fig.savefig(ROOT / 'docs' / 'desbalance_pagevalues.png', dpi=120)
plt.show()
```
